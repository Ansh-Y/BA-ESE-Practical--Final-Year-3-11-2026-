```python
import numpy as np
import matplotlib.pyplot as plt

# Input 2D Data Points
data = np.array([
    [2, 3],
    [3, 4],
    [4, 3],
    [10, 11],
    [11, 12],
    [12, 11]
])

# Number of clusters
k = 2

# Initial centroids
centroids = np.array([
    [2, 3],
    [10, 11]
])

# K-Means Algorithm
for i in range(10):

    # Calculate Euclidean distance
    distances = np.linalg.norm(
        data[:, np.newaxis] - centroids,
        axis=2
    )

    # Assign points to nearest centroid
    labels = np.argmin(distances, axis=1)

    # Calculate new centroids
    new_centroids = np.array([
        data[labels == j].mean(axis=0)
        for j in range(k)
    ])

    # Check convergence
    if np.allclose(centroids, new_centroids):
        break

    centroids = new_centroids

# Display Results
print("Cluster Labels:")
print(labels)

print("\nFinal Centroids:")
print(centroids)

# Plot clusters
plt.scatter(data[:, 0], data[:, 1], c=labels)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color='red',
    marker='X',
    s=200,
    label='Centroids'
)

plt.title("K-Means Clustering for 2D Data")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```
